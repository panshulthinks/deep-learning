```python
import os
import shutil
import random

import os
import shutil
import random

# Paths
dataset_path = r"C:\Users\Panshul\Desktop\panshul\download\meme_dataset"
train_path = os.path.join(dataset_path, "train")
test_path = os.path.join(dataset_path, "test")

# Create test folder structure
for cls in os.listdir(train_path):
    os.makedirs(os.path.join(test_path, cls), exist_ok=True)

# Split ratio
test_ratio = 0.2

# Move files to test folder
for cls in os.listdir(train_path):
    cls_train_path = os.path.join(train_path, cls)
    cls_test_path = os.path.join(test_path, cls)
    
    images = os.listdir(cls_train_path)
    num_test = int(len(images) * test_ratio)
    
    test_images = random.sample(images, num_test)
    
    for img in test_images:
        src = os.path.join(cls_train_path, img)
        dst = os.path.join(cls_test_path, img)
        shutil.move(src, dst)

print("Dataset split into train and test successfully!")
```

```python
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# Device
device = "cuda" if torch.cuda.is_available() else "cpu"

# Paths
train_dir = r"C:\Users\Panshul\Desktop\panshul\download\meme_dataset\train"
test_dir  = r"C:\Users\Panshul\Desktop\panshul\download\meme_dataset\test"

# Transformations
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

# Load datasets
train_dataset = datasets.ImageFolder(train_dir, transform=train_transform)
test_dataset  = datasets.ImageFolder(test_dir, transform=test_transform)

# Dataloaders
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader  = DataLoader(test_dataset, batch_size=32, shuffle=False)

print(f"Train samples: {len(train_dataset)}, Test samples: {len(test_dataset)}")
```

```python
import torch.nn as nn
import torchvision.models as models
import torch.optim as optim

model = models.resnet18(pretrained=True)

num_ftrs = model.fc.in_features
model.fc = nn.Linear(num_ftrs, 2)  # 2 classes: Meme vs No_Meme
model = model.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=1e-4)
```

```python
num_epochs = 5  

for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0

    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()
    
    print(f"Epoch [{epoch+1}/{num_epochs}], Loss: {running_loss/len(train_loader):.4f}")
```

```python
model.eval()
correct = 0
total = 0

with torch.no_grad():
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

print(f"Test Accuracy: {100 * correct / total:.2f}%")
```

```python
import torch
import torch.nn.functional as F
import numpy as np
import cv2
import matplotlib.pyplot as plt

def show_gradcam(model, image, class_idx, device):
    model.eval()
    image = image.unsqueeze(0).to(device)
    
    # ----------------------------
    # 1. Forward hook to get features
    # ----------------------------
    features = []
    def forward_hook(module, input, output):
        features.append(output)
    
    handle = model.layer4.register_forward_hook(forward_hook)
    
    # Forward pass
    output = model(image)
    pred_class = output.argmax(dim=1).item()
    
    # ----------------------------
    # 2. Backward pass to get gradients
    # ----------------------------
    model.zero_grad()
    class_score = output[0, class_idx]
    class_score.backward()
    
    # Gradients of layer4 feature map
    gradients = features[0].grad if features[0].grad is not None else None
    if gradients is None:
        # Register hook on feature to save gradients
        gradients = []
        def backward_hook(module, grad_in, grad_out):
            gradients.append(grad_out[0])
        features[0].register_hook(backward_hook)
        class_score.backward()
        gradients = gradients[0]
    
    # ----------------------------
    # 3. Compute Grad-CAM
    # ----------------------------
    pooled_gradients = torch.mean(gradients, dim=[0,2,3])
    feature_map = features[0][0]
    for i in range(feature_map.shape[0]):
        feature_map[i, :, :] *= pooled_gradients[i]
    
    heatmap = feature_map.mean(dim=0).cpu().detach().numpy()
    heatmap = np.maximum(heatmap, 0)
    heatmap /= np.max(heatmap)
    
    # ----------------------------
    # 4. Overlay heatmap on image
    # ----------------------------
    img = image[0].cpu().permute(1,2,0).numpy()
    # unnormalize if image was normalized for ImageNet
    img = img * np.array([0.229,0.224,0.225]) + np.array([0.485,0.456,0.406])
    img = np.clip(img, 0, 1)
    
    heatmap = cv2.resize(heatmap, (img.shape[1], img.shape[0]))
    heatmap = cv2.applyColorMap(np.uint8(255*heatmap), cv2.COLORMAP_JET)
    superimposed_img = cv2.addWeighted(np.uint8(img*255), 0.6, heatmap, 0.4, 0)
    
    plt.imshow(superimposed_img)
    plt.axis('off')
    plt.show()
    
    # Remove hook
    handle.remove()
```
