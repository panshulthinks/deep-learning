```python
from datasets import load_dataset

dataset = load_dataset("go_emotions")

print(dataset)
```

```python
print(dataset['train'][0])
```

```python
labels = dataset['train'].features['labels'].feature.names
print(f"Total emotions: {len(labels)}")
print(labels)
```

```python
import matplotlib.pyplot as plt
from collections import Counter
```

```python
all_labels = [label for labels_list in dataset['train']['labels'] for label in labels_list]
counts = Counter(all_labels)

plt.figure(figsize=(12, 5))
plt.bar(range(len(labels)), [counts[i] for i in range(len(labels))])
plt.xticks(range(len(labels)), labels, rotation=90)
plt.title("Emotion Label Distribution (Train Set)")
plt.show()
```
