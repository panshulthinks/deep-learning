```python
import re
```

```python
chat1 = 'codebasics: you ask lot of questions 1234567890, abc@xyz.com'
chat2 = 'codebasics: here it is (123)-567-8912, ab34@xyz.com'
chat3 = 'codebasics: yes, phone:1234567890 email: xyz@xyz.com'
```

```python
pattern = '\d{10}|\(\d{3}\)-\d{3}-\d{4}'
matches = re.findall(pattern, chat1 + " " + chat2 + " " + chat3)
matches
```

```python
pattern = '[a-z0-9A-Z_]*@[a-zA-Z0-9]*\.[a-z]*'
matches = re.findall(pattern, chat1 + " " +chat2 + " " + chat3)
matches
```
