```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.ensemble import RandomForestClassifier
from  sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv("IMDB Dataset.csv")
df
```

```python
df.shape
```

```python
df.review.value_counts()
```

```python
df['Category'] = df['sentiment'].apply(lambda x:1 if x =='positive' else 0)
```

```python
df.shape
```

```python
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.review, df.sentiment, 
                                                    test_size = 0.2)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
v = CountVectorizer()

X_train_cv = v.fit_transform(X_train.values)
X_train_cv
```

```python
v.vocabulary_
```

```python
clf = Pipeline([
    ('vectorizer', CountVectorizer()),                                                    
    ('random_forest', (RandomForestClassifier(n_estimators=50, criterion='entropy')))      
])
```

```python
clf.fit(X_train, y_train)
```

```python
y_pred = clf.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
```

```python
clf = Pipeline([
                
     ('vectorizer', CountVectorizer()),   
      ('KNN', (KNeighborsClassifier(n_neighbors=10, metric = 'euclidean')))   #using the KNN classifier with 10 neighbors 
])
```

```python
clf.fit(X_train, y_train)
```

```python
y_pred = clf.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
```

```python
clf = Pipeline([
                
     ('vectorizer', CountVectorizer()),   
      ('Multi NB', MultinomialNB())   #using the Multinomial Naive Bayes classifier 
])
```

```python
clf.fit(X_train, y_train)
```

```python
y_pred = clf.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
```
