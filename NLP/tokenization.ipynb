```python
nlp = spacy.blank("hi")

doc = nlp("भाई कृपया मेरे 500 ₹ लौटा दो जो तुम्हें मुझ पर बकाया हैं")
for token in doc:
    print(token, token.is_currency, token.like_num)
```

```python
doc = nlp("gimme double cheese extra large healthy pizza")

tokens = [token.text for token in doc]
tokens
```

```python
from spacy.symbols import ORTH

nlp.tokenizer.add_special_case("gimme",[
    {ORTH: "gim"},
    {ORTH: "me"}
])

doc = nlp("gimme double cheese extra large healthy pizza")

tokens = [token.text for token in doc]
tokens
```

```python
doc = nlp("Dr. Strange loves pav bhaji of mumbai. Hulk loves chat of delhi")
for sentence in doc.sents:
    print(sentence)
```

```python
transactions = nlp("Tony gave two $ to Peter, Bruce gave 500 € to Steve")

for token in transactions:
    if token.like_num and transactions[token.i+1].is_currency:
        print(token.text, transactions[token.i+1].text)
```
