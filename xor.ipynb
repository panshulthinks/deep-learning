### implementing xor from scratch without using pytorch/tensor

### XOR is not linearly seperable thats why we would need a hidden layer

```python
import numpy as np
```

```python
# XOR dataset
x = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])
```

### lets make input layers with 2 neurons
### the hideen layer with 2 neurons(ReLU)
### output layer with 1 neuron(Sigmoid)

```python
input_dim = 2
hidden_dim = 4
output_dim = 1
```

### lets initalize paramters

```python
w1 = np.random.randn(input_dim, hidden_dim)
b1 = np.zeros((1, hidden_dim))
w2 = np.random.randn(hidden_dim, output_dim)
b2 = np.zeros((1, output_dim))
```

### lets define our activation function

```python
def sigmoid(x):
    return 1/(1+ np.exp(-x))
def sigmoid_derivative(x):
    return x*(1-x)
def relu(x):
    return np.maximum(0, x)
def relu_derivative(x):
    return np.where(x > 0, 1, 0)
```

### now we have to do training (forward and backward)

```python
lr = 1
epochs = 10000

for i in range(epochs):
    z1 = np.dot(x, w1) + b1
    a1 = relu(z1)
    
    z2 = np.dot(a1, w2) + b2
    a2 = sigmoid(z2)
    
    loss = np.mean((y - a2)**2)
    
    da2 = a2 - y
    dz2 = da2 * sigmoid_derivative(a2)
    dw2 = np.dot(a1.T, dz2)
    db2 = np.sum(dz2, axis=0, keepdims=True)
    
    da1 = np.dot(dz2, w2.T)
    dz1 = da1 * relu_derivative(z1)
    dw1 = np.dot(x.T, dz1)
    db1 = np.sum(dz1, axis=0, keepdims=True)
    
    w2 -= lr * dw2
    b2 -= lr * db2
    w1 -= lr * dw1
    b1 -= lr * db1

    if i % 1000 == 0:
        print(f"epoch {i}, loss: {loss:.4f}")
```
